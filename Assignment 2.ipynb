```python

# 1(a) Create a list with five integer elements
age_list = [24, 25, 26, 27, 28]

# Print the age list
print("Age List:", age_list)

# 1(b) Create a list with five string elements
name_list = ["Arun", "Binu", "Charan", "Deepa", "Esha"]

# Print the name list
print("Name List:", name_list)
```

```python

# 2(a) Append the string "Yazhini" to name_list
name_list.append("Yazhini")
print("After append:", name_list)

# 2(b) Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After insert:", age_list)

# 2(c) Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After remove:", name_list)

# 2(d) Remove the last element from age_list
age_list.pop()
print("After pop:", age_list)

# 2(e) Extend age_list with additional ages
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# 2(f) Sort age_list in descending order
age_list.sort(reverse=True)
print("Descending order:", age_list)

# 2(g) Find maximum, minimum and sum of ages
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python

# 3(a) Print the first element of name_list
print("First element:", name_list[0])

# 3(b) Print the last element of name_list
print("Last element:", name_list[-1])

# 3(c) Print elements from index 2 to index 4
# Python slicing excludes the ending index, so use 2:5
print("Elements from index 2 to 4:", name_list[2:5])

# 3(d) Print the elements in reverse order
print("Reverse order:", name_list[::-1])
```

```python

# 2(a) Create a dictionary containing five students and their marks
student_marks = {
    "Arun": 75,
    "Binu": 88,
    "Charan": 65,
    "Deepa": 92,
    "Esha": 79
}

# Display the dictionary
print("Student Marks:", student_marks)

# 2(b) Access and print the mark of a specific student
print("Arun's mark:", student_marks["Arun"])
```

```python

# 2(c) Add a new student named Janani with mark 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# 2(d) Update an existing student's mark to 82
student_marks["Binu"] = 82
print("After updating Binu's mark:", student_marks)
```

```python

# 2(e) Print all student names using keys()
print("Student Names:", student_marks.keys())

# Print all marks using values()
print("Student Marks:", student_marks.values())

# Print all key-value pairs using items()
print("Student and Marks:", student_marks.items())
```

```python

# 3(a) Create a set containing the given vowel values
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])

# Print the set
print("My Set:", my_set)

# Explanation:
# A set stores only unique elements.
# Duplicate values like 'a' and 'i' are removed automatically.
# The order of elements in a set is not guaranteed.
```

```python

# 3(b) Attempt to change the element at index 4
# Sets do not support indexing or item assignment

try:
    my_set[4] = 's'
except TypeError as error:
    print("Error:", error)
    print("Explanation: Sets are unordered and do not support indexing or item assignment.")
```

```python

# 3(c) Create the two given sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Print both sets
print("Set 1:", set1)
print("Set 2:", set2)

# 3(c) Create the two given sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Print both sets
print("Set 1:", set1)
print("Set 2:", set2)
```

```python

# 3(d) Find the union of set1 and set2
# Union contains all unique elements from both sets
print("Union:", set1.union(set2))

# Find the intersection of set1 and set2
# Intersection contains only common elements
print("Intersection:", set1.intersection(set2))
```

```python

# Ask the user to enter a score between 0 and 10
score = float(input("Enter your score (0 to 10): "))

# Check whether the score is within the valid range
if score < 0 or score > 10:
    print("Invalid score! Please enter a score from 0 to 10.")

# Categorise the performance using if, elif and else
elif score > 7:
    print("Above Average: Excellent work! Keep it up.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
